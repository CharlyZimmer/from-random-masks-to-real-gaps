```python
import sys, os
sys.path.insert(0, os.environ.get('GAP_AWARE_ROOT', '../../../..'))

from patchfill.dataset_analysis import plot_global_map

import xarray as xr
import cmocean
```

```python
path = os.environ.get('ESDC_DATA_DIR')
var = 'land_surface_temperature'
time_slice = '2011-09-18'
val_range = (273.15, 323.15)
cmap = cmocean.cm.matter_r

# Set to a specific filename to pick a particular run, e.g.:
#   no_gaps_zarr = 'no_gaps_2005-06-01_2005-07-31.zarr'
# Leave as None to auto-select (first match alphabetically).
no_gaps_zarr = None
```

```python
import glob, os

if no_gaps_zarr is None:
    candidates = sorted(glob.glob(os.path.join(path, 'no_gaps_*.zarr')))
    if not candidates:
        raise FileNotFoundError(f'No no_gaps_*.zarr found in {path}')
    no_gaps_zarr = os.path.basename(candidates[0])
    if len(candidates) > 1:
        print(f'Multiple no_gaps zarrs found — using {no_gaps_zarr!r} (first alphabetically).')
        print('Set no_gaps_zarr in the config cell to use a different one:')
        for c in candidates:
            print(f'  {os.path.basename(c)}')
    else:
        print(f'Using {no_gaps_zarr!r}')

ds = xr.open_zarr(path + "/" + no_gaps_zarr)
```

```python
plot_global_map(ds, var, cmap, val_range, time=time_slice)
```

```python
plot_global_map(ds, var + '_no_gaps', cmap, val_range, time=time_slice)
```
